```python
import fasttext

# Create data automatically (no manual file)
sentences = [
    "i love machine learning",
    "machine learning is amazing",
    "deep learning is powerful",
    "nlp is interesting",
    "python is easy",
    "ai is the future",
    "data science is growing",
    "coding is fun"
]

# Write internally (you don't need to create manually)
with open("temp.txt", "w") as f:
    for _ in range(1000):   # repeat to increase data
        for s in sentences:
            f.write(s + "\n")

# Train model
model = fasttext.train_unsupervised("temp.txt", model="skipgram", minCount=1)

# Use embedding
print(model.get_word_vector("machine"))
print(model.get_nearest_neighbors("machine"))
```

```python
import re
```

```python
def remove_html_tags(text):
  pattern = re.compile("<.*?>")
  return pattern.sub(r"",text)
```

```python
text1 = "my name is <ronak>"
remove_html_tags(text1)
```

```python
text = remove_html_tags(text)
```

```python
text
```

```python
def remove_urls(text):
  pattern = re.compile(r"https?://\S+|www\.\S+")
  return pattern.sub(r"", text)
```

```python
text = remove_urls(text)
```

```python
import string
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char , "")
  return text
```

```python
text = remove_punc(text)
```
