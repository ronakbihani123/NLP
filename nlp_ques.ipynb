```python
# why do we remove stopwords in nlp ? can removing stopwords ever reduce model performance ?
```

```python
# stopwords are common words like 'is', 'and' , 'in' , 'this'
# we remove them because they don't add much meaning to the sentence

# Reasons--->
# 1. Reduce Noise (unnecessary data):
# this words appear in almost every sentence .
# They don’t help the model understand the topic.
# 👉 Example:
# "The movie is very good"
# → Important word = good, not the, is

# 2.Faster processing:
# Fewer words → smaller data → faster training
# Helps in improving efficiency

# 3.Better focus on important words
# Model focuses on meaningful words like “good”, “bad”, “product”

# 4.Useful for traditional models
# Models like Bag of Words, TF-IDF work better after removing stopwords

# Can removing stopwords reduce model performance?

# 👉 Yes, in some cases it can actually hurt performance.
# In sentiment analysis

# Words like “not”, “very”, “too” are important
# Removing them can change the emotion of the sentence

# When grammar matters
# Tasks like translation, chatbot, text generation need full sentences
```

```python
# when should u use stemming vs lemmatization
```

```python
# 🔹 Use Stemming when:
# You want fast processing
# Accuracy is not very critical
# Working with large datasets
# Example tasks: search engines, basic text classification

# 👉 It just cuts words to root form (may be incorrect)
# Example: “running” → “run”, “studies” → “studi”

# 🔹 Use Lemmatization when:
# You need meaningful, correct words
# Accuracy is important
# Working on NLP tasks like chatbot, sentiment analysis

# 👉 It converts words to dictionary form (lemma)
# Example: “running” → “run”, “better” → “good”
```

```python
# why are word embeddings better than one hot encoding ?
```

```python
# 1. Capture meaning (semantics)

# 👉 Example:

# “king” ≈ “queen”
# “dog” ≈ “cat”
# 🔹 2. Dense and compact vectors
# Small size (e.g., 100, 300 dimensions)
# More efficient
# 🔹 3. Understand relationships

# 👉 Famous example:

# king - man + woman ≈ queen
# 🔹 4. Improve model performance
# Better for tasks like:
# Sentiment analysis
# Chatbots
# Translation
# 🧠 Interview one-liner:

# 👉 “Word embeddings are better than one-hot encoding because they capture semantic meaning and
#  relationships between words using dense vectors, whereas one-hot vectors are sparse and carry no
#   contextual information.”
```
