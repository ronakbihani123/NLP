```python
# difference btw rule based and ml based ner
```

```python
# # # # Uses predefined rules and patterns to find entities.

# # # # 👉 These rules are written manually.
# # # Words starting with capital letters → Person
# # # Words after “Mr.” → Person
# # # Words ending with “Ltd”, “Inc” → Organization

# ✅ Example:

# Sentence: "Mr. Sharma works at TCS"

# "Mr. Sharma" → Person (rule: starts with Mr.)
# "TCS" → Organization (rule-based list)

# # Uses machine learning models trained on labeled data.

# # 👉 The model learns patterns automatically.

# Sentence: "Elon Musk founded SpaceX"
# Model learns:

# "Elon Musk" → Person
# "SpaceX" → Organization

# # Often part of Natural Language Processing
```

```python
# what is bio tagging in ner ?
```

```python
# BIO tagging is a way to label each word in a sentence so a model can understand where an entity starts, continues, and ends.
```

```python
# ✨ Simple example

# Sentence:
# 👉 "Virat Kohli lives in Delhi"

# BIO tagging will look like this:

# Word	Tag
# Virat	B-PER
# Kohli	I-PER
# lives	O
# in	O
# Delhi	B-LOC
```

```python
# difference btw ner and text classification
```

```python
# Feature	                                 NER (Named Entity Recognition)	                              Text Classification
# Purpose	                                 Identify entities in text	                                  Assign category/label to whole text
# Output	                                 Labels for specific words/tokens	                            One label for entire sentence/document
# Focus	                                   Important entities	                                          Overall meaning/topic
# Level	                                   Token-level task	                                            Sentence/document-level task
```
