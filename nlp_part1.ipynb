```python
import re
def remove_html_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r"", text)
```

```python
text = "<html><body><p> Movie1 </p><p> click here to <a href = 'http://google.com'> download</a></p?> "
remove_html_tags(text)
```

```python
# now we convert tags from imdb datasets
df['review'] = df['review'].apply(remove_html_tags)
```

```python
def remove_urls(text):
  pattern = re.compile(r"https?://\S+|www\.\S+")
  return pattern.sub(r"", text)
```

```python
# remove punctuation
```

```python
import string
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char , "")
  return text
```

```python
text = 'string. with. Punctuation?'
```

```python
remove_punc(text)
```

```python
df['review'] = df['review'].apply(remove_punc)
```

```python
print(remove_punc(text))
```

```python
# if we have a large amount of data then we use this method
def remove_punc1(text):
  return text.translate(str.maketrans('' , '' , exclude))
```

```python
import time

start = time.time()
print(remove_punc1(text))
time2 = time.time() - start
print(time2)
```

```python
# Chatword Treatment
```

```python
chat_words = {'u2': 'you too' , 'BBL':'Be Back Later' , 'GAL': 'Get A Life' }
```

```python
chat_words
```

```python
def chat_conversion(text):
  new_text = []
  for w in text.split():
    if w.upper() in chat_words:
      new_text.append(chat_words[w.upper()])
    else:
      new_text.append(w)

  return "".join(new_text)
```

```python
chat_conversion('BBL')
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
STOPWORDS = set(stopwords.words('english'))
```

```python
STOPWORDS
```

```python
# MANUALLY WORK
def remove_stopwords(text):
  new_text = []
  for word in text.split():
    if word in set(stopwords.words('english')):
      new_text.append('')
    else:
      new_text.append(word)

  x= new_text[:]
  new_text.clear()
  return "".join(x)

remove_stopwords('probably my all-time favourite movie , a story of selflessness , sacrifice and dedication to a noble cause')
```
