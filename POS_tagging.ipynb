```python
# POS Tagging in NLP

# POS tagging means part of speech tagging
# it is an NLP technique where each word in a sentence is assigned its
# grammatical role such as :
# NOUN
# verb
# adjective
# adverb
# pronoun
# preposition
# conjunction
# interjection
```

```python
# # why pos tagging is important ?

# # pos tagging is used in :
# # text classiffication
# # chatbots
# # machine translation
# # speech recognition
# # namer entity recognition
# # sentiment analysis
# grammar checking
```

```python
# from typing import NoReturn
# common pos tags

# POS Tag      |      Meaning

# NN         |      Noun
# NNS         |     plural noun
# NNP        |     proper noun
# VB          |  verb
# VBD        |  verb past tense
# VBG           | verb gerund
# jj          |  adjective
# rb           | adverb
# PRP          | Pronoun
# IN             | preposition
```

```python
# from IPython.core.inputtransformer2 import TokenTransformBase
# working process of pos tagging :::::::
# 1. sentence is divided into Token
# 2 . NLP model checks grammar / context
# 3. appropriate POs tag is assigned to each word
```

```python
import spacy

# Load english model
nlp = spacy.load("en_core_web_sm")

# input text
doc = nlp("Sourabh is learning NLP")

# POS tagging
for token in doc:
  print(token.text, "-->", token.pos_)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("Sourabh is learning NLP")

for token in doc:
  print("word:" , token.text,
        "| POS:" , token.pos_,
        "|TAG:" , token.tag_)
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/IMDB Dataset.csv")
df.head()
```

```python
df["review"] = df["review"].str.lower()
```

```python
import re
def remove_html_tags(text):
  pattern = re.compile("<.*?>")
  return pattern.sub(r"", text)
```

```python
df["review"] = df["review"].apply(remove_html_tags)
```

```python
df
```

```python
def remove_urls(text):
  pattern  = re.compile(r"https://\S+|www\. \S+")
  return pattern.sub(r"", text)
```

```python
text = "For notebook click https://www.deture.com/delta/note1234acc"
remove_urls(text)
```

```python
df["review"] = df["review"].apply(remove_urls)
```

```python
import string
string.punctuation
```

```python
exclude= string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char , "")
  return text
```

```python
text = 'string. with. Punctuation?'
```

```python
remove_punc(text)
```

```python
df["review"] = df["review"].apply(remove_punc)
```

```python
df
```

```python
paragraph = df["review"][3]
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp(paragraph)

for token in doc:
  print(token.text , "----->" , token.pos_)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(paragraph)
for ent in doc.ents:
  print(ent.text , "------>" , ent.label_)
```
